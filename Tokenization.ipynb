```python
from transformers import AutoTokenizer
```

```python
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
```

```python
text = "kalyani is very good girl!"
```

```python
tokens = tokenizer.tokenize(text)
```

```python
print("Tokens:",tokens)
```

```python
#encode
encoded = tokenizer(text)
```

```python
print("\nInput IDs",encoded["input_ids"])
```

```python
print("Length of the tokens:",len(tokens))
```

```python
vocab = tokenizer.get_vocab()
```

```python
print("Length of the vocabularies:",len(vocab))
```

```python
import tiktoken
```

```python


encoding = tiktoken.encoding_for_model("gpt-4o")
```

```python
text = "kalyani is very goog girl"
```

```python
tokens = encoding.encode(text)
```

```python
print("Tokens:",tokens)
```

```python
print("\nIndividual tokens:")
for token in tokens:
    print(encoding.decode([token]), end=" | ")
```
