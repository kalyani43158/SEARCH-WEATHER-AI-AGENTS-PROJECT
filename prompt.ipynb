```python
import os
from groq import Groq

client = Groq(api_key=os.environ.get("GROQ_API_KEY"))

prompt = """
Analyze the power line inspection result.
Output TRUE if a fault or obstacle is detected, otherwise FALSE.

Example 1:
Input: Power line is healthy and no obstacle is nearby.
Output: FALSE

Example 2:
Input: Cracked insulator detected.
Output: TRUE

Example 3:
Input: Tree branch is very close to the power line.
Output: TRUE

Example 4:
Input: Power line is normal with no visible damage.
Output: FALSE

Input: Abnormal heating detected on the power line.
Output:
"""

response = client.chat.completions.create(
    messages=[{"role": "user", "content": prompt}],
    model="llama-3.3-70b-versatile"
)

print(response.choices[0].message.content)
```

```python
import os
from groq import Groq

client = Groq(
    api_key=os.environ.get("GROQ_API_KEY"),
)

prompt = """
Solve the following problem using:
Step 1 → Step 2 → Step 3 → Final Answer.

Question:
A drone battery has 2400 mAh capacity.
During inspection, it uses 15% of the battery for takeoff,
25% for power-line scanning, and 20% for return flight.
How much battery capacity remains?

Step 1:
Calculate the total percentage used.

Step 2:
Calculate the battery capacity used.

Step 3:
Subtract the used capacity from the original capacity.

Final Answer:
"""

chat_completion = client.chat.completions.create(
    messages=[
        {
            "role": "user",
            "content": prompt
        }
    ],
    model="llama-3.3-70b-versatile",
)

print(chat_completion.choices[0].message.content)
```
