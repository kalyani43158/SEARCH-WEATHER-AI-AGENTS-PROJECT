```python
model.invoke("what is current wheather in ongole")
```

```python
from langchain.tools import tool
@tool
def get_wheather(location:str)->str:
    """ get a wheather at a location"""
    return f"it's sunny{location}"
model_with_tool = model.bind_tools([get_wheather])
```

```python
response = model_with_tool.invoke(
    "What's the weather like in Ongole?"
)
print(response)
```

```python
for tool_call in response.tool_calls:
    print(f"Tool: {tool_call['name']}")
    print(f"Args: {tool_call['args']}")
```
